Interactive preparation covering **Dynamical States, Nonlinear Feedback, Shannon Entropy, Network Adjacency, and Lagged Time Series**.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

def run_primer_sandbox(r_growth=3.75, lag_tau=2, p_coin=0.5):
    # 1. Nonlinear Dynamics & Feedback: Logistic Map x_{t+1} = r * x_t * (1 - x_t)
    n_steps = 120
    x = np.zeros(n_steps)
    x[0] = 0.21
    for t in range(n_steps - 1):
        x[t + 1] = r_growth * x[t] * (1.0 - x[t])

    # 2. Lagged Time Series Embedding (X_{t - tau} vs X_t)
    x_past = x[:-lag_tau]
    x_curr = x[lag_tau:]
    cov_matrix = np.cov(x_past, x_curr)
    corr_val = np.corrcoef(x_past, x_curr)[0, 1]

    # 3. Logarithm & Shannon Entropy H(X) for Binary Source
    p_vals = np.linspace(0.001, 0.999, 200)
    h_curve = -(p_vals * np.log2(p_vals) + (1.0 - p_vals) * np.log2(1.0 - p_vals))
    h_curr = -(p_coin * np.log2(p_coin) + (1.0 - p_coin) * np.log2(1.0 - p_coin))

    fig, axes = plt.subplots(1, 3, figsize=(14.5, 4.2))
    fig.patch.set_facecolor('#F8FAFC')

    # Panel 1: State Trajectory
    axes[0].set_facecolor('#F8FAFC')
    axes[0].plot(x[:60], color='#2563EB', linewidth=1.8, marker='o', markersize=3)
    axes[0].set_title(f'1. Nonlinear State Evolution (r={r_growth:.2f})', fontsize=11, fontweight='bold')
    axes[0].set_xlabel('Time Step t')
    axes[0].set_ylabel('State x(t)')
    axes[0].set_ylim(0, 1.05)
    axes[0].grid(True, linestyle='--', alpha=0.4)

    # Panel 2: Lagged Phase Portrait
    axes[1].set_facecolor('#F8FAFC')
    axes[1].scatter(x_past, x_curr, color='#D97706', alpha=0.8, edgecolors='#0F172A', s=32)
    axes[1].set_title(f'2. Lagged Series (tau={lag_tau}, r={corr_val:.2f})', fontsize=11, fontweight='bold')
    axes[1].set_xlabel(f'Past State x(t - {lag_tau})')
    axes[1].set_ylabel('Current State x(t)')
    axes[1].set_xlim(0, 1.05)
    axes[1].set_ylim(0, 1.05)
    axes[1].grid(True, linestyle='--', alpha=0.4)

    # Panel 3: Binary Shannon Entropy
    axes[2].set_facecolor('#F8FAFC')
    axes[2].plot(p_vals, h_curve, color='#0D9488', linewidth=2.2)
    axes[2].scatter([p_coin], [h_curr], color='#E11D48', s=80, zorder=5)
    axes[2].set_title(f'3. Shannon Entropy H(X) = {h_curr:.3f} bits', fontsize=11, fontweight='bold')
    axes[2].set_xlabel('Event Probability P(X=1)')
    axes[2].set_ylabel('Uncertainty (Bits)')
    axes[2].grid(True, linestyle='--', alpha=0.4)

    plt.tight_layout()
    plt.show()
    print(f'Lagged Covariance Cov(X_{{t-{lag_tau}}}, X_t) = {cov_matrix[0, 1]:.4f} | Shannon Entropy = {h_curr:.4f} bits')

widgets.interact(
    run_primer_sandbox,
    r_growth=widgets.FloatSlider(min=2.5, max=4.0, step=0.05, value=3.75, description='Feedback r:'),
    lag_tau=widgets.IntSlider(min=1, max=8, step=1, value=1, description='Time Lag tau:'),
    p_coin=widgets.FloatSlider(min=0.05, max=0.95, step=0.05, value=0.5, description='Prob P(X=1):'),
);
```
